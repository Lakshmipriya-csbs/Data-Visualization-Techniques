```python
import pandas as pd

data = {
    "ID": [201, 202, 203, 204, 205, 206],
    "Name": ["Meena", "Vikram", "Anjali", "Suresh", "Neha", "Rahul"],
    "Department": ["ECE", "EEE", "MECH", "ECE", "CIVIL", "EEE"],
    "Marks": [68, 82, 74, 91, 77, 86],
    "Attendance": [84, 93, 89, 97, 81, 90]
}

df = pd.DataFrame(data)

print(df.groupby("Department")["Marks"].mean())
print(df.groupby("Department")["Attendance"].mean())
```
