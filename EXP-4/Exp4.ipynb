```python
import pandas as pd
df = pd.read_csv("231401057.csv")
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
high_streams = df[df["Total Streams (in millions)"] > 50000]
print(high_streams)
```

```python
solo_artists = df[df[" Artist Type"] == "Solo"]
print(solo_artists)
```

```python
sorted_df = df.sort_values(
    by="Total Streams (in millions)",
    ascending=False
)
print(sorted_df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

sorted_df = df.sort_values(by="Total Streams (in millions)", ascending=False)

plt.figure(figsize=(12,5))
plt.bar(range(len(sorted_df)), sorted_df["Total Streams (in millions)"])
plt.title("Total Streams Ranking")
plt.xlabel("Artists (Sorted)")
plt.ylabel("Total Streams (in millions)")
plt.show()
```

```python
print("Mean =", df["Total Streams (in millions)"].mean())
print("Median =", df["Total Streams (in millions)"].median())
print("Mode =",df["Total Streams (in millions)"].mode())
print("Standard Deviation =", df["Total Streams (in millions)"].std())
```
