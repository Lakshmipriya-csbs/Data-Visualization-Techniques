```python
import matplotlib.pyplot as plt

departments = ["HR", "IT", "Finance", "Sales"]
employees = [25, 40, 20, 15]

plt.pie(employees, labels=departments, autopct="%1.1f%%")

plt.title("Department-wise Employee Distribution")

plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 68, 70, 72, 75,
         78, 80, 82, 85, 88, 90, 92, 95]

plt.hist(marks, bins=5)

plt.title("Student Marks Distribution")
plt.xlabel("Marks Obtained")
plt.ylabel("Number of Students")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["January", "February", "March", "April", "May", "June"]
sales = [120, 150, 180, 170, 220, 250]

plt.bar(months, sales)

plt.title("Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales Amount")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["January", "February", "March", "April", "May", "June"]
sales = [120, 150, 180, 170, 220, 250]

plt.plot(months, sales, marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales Amount")

plt.show()
```
