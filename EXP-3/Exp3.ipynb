```python
import pandas as pd
df = pd.read_csv("231401057.csv")
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df["Debut Year"] = df["Debut Year"].fillna(df["Debut Year"].mean())
df["Total Streams (in millions)"] = df["Total Streams (in millions)"].fillna(df["Total Streams (in millions)"].mean())

print(df)
```

```python

print(df.duplicated())

df.drop_duplicates(inplace=True)

print(df)
```

```python
Q1 = df['Total Streams (in millions)'].quantile(0.25)
Q3 = df['Total Streams (in millions)'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Total Streams (in millions)'] < lower) |
              (df['Total Streams (in millions)'] > upper)]

print(outliers)
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['Total Streams (in millions)']] = scaler.fit_transform(df[['Total Streams (in millions)']])

print(df[['Total Streams (in millions)']])
```
