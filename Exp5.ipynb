```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("231401057.csv")

data = df.head()

plt.plot(data["Debut Year"], data["Total Streams (in millions)"])

plt.title("Debut Year vs Total Streams")
plt.xlabel("Debut Year")
plt.ylabel("Total Streams")

plt.show()
```

```python
plt.bar(data["Debut Year"], data["Total Streams (in millions)"])

plt.title("Debut Year vs Total Streams")
plt.xlabel("Debut Year")
plt.ylabel("Total Streams")

plt.show()
```

```python
artist_type_count = data[" Artist Type"].value_counts()

plt.pie(
    artist_type_count,
    labels=artist_type_count.index,
    autopct='%1.1f%%'
)

plt.title("Artist Type Distribution")
plt.show()
```

```python
plt.boxplot(data["Total Streams (in millions)"])

plt.title("Total Streams")
plt.ylabel("Total Streams")

plt.show()
```
